## Theory Notes

# 1. Classes and Objects

Object-Oriented Programming (OOP) is a programming paradigm that organizes code using **classes** and **objects**. It helps create reusable, scalable, and modular programs.

### Class

A class is a blueprint or template used to create objects. It defines attributes (variables) and methods (functions).

### Object

An object is an instance of a class that contains actual values and can access methods defined in the class.

### Syntax

```python
class ClassName:
    pass

obj = ClassName()
```

### Example

```python
class Student:
    name = "Aditya"

student1 = Student()

print(student1.name)
```

### Output

```text
Aditya
```

### Advantages

* Reusability
* Better code organization
* Easy maintenance
* Real-world modeling

---

# 2. Constructors (`__init__`)

A constructor is a special method used to initialize object data automatically when an object is created.

In Python, the constructor method is called `__init__()`.

### Syntax

```python
class ClassName:
    def __init__(self):
        pass
```

### Example

```python
class Student:

    def __init__(self, name):
        self.name = name

student1 = Student("Aditya")

print(student1.name)
```

### Output

```text
Aditya
```

### Importance of Constructor

* Initializes object values
* Executes automatically
* Reduces repetitive code

---

# 3. Instance Variables and Methods

## Instance Variables

Instance variables are unique variables for each object. They are defined using `self`.

### Example

```python
class Student:

    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

student1 = Student("Aditya", 90)

print(student1.name)
print(student1.marks)
```

### Output

```text
Aditya
90
```

## Instance Methods

Methods that work with object data are called instance methods.

### Example

```python
class Student:

    def __init__(self, name):
        self.name = name

    def display(self):
        print("Student Name:", self.name)

student1 = Student("Aditya")

student1.display()
```

### Output

```text
Student Name: Aditya
```

---

# 4. Class Variables

Class variables are shared by all objects of a class.

### Example

```python
class Student:

    school_name = "ABC School"

    def __init__(self, name):
        self.name = name

student1 = Student("Aditya")
student2 = Student("Rahul")

print(student1.school_name)
print(student2.school_name)
```

### Output

```text
ABC School
ABC School
```

### Difference Between Class Variable and Instance Variable

| Class Variable        | Instance Variable          |
| --------------------- | -------------------------- |
| Shared by all objects | Unique for each object     |
| Defined inside class  | Defined inside constructor |

---

# 5. Encapsulation

Encapsulation means hiding data and controlling access using methods.

It protects object data from accidental modification.

Python uses:

* Public Members
* Protected Members
* Private Members

### Example

```python
class Bank:

    def __init__(self, balance):
        self.__balance = balance

    def get_balance(self):
        return self.__balance

account = Bank(5000)

print(account.get_balance())
```

### Output

```text
5000
```

### Benefits

* Data security
* Controlled access
* Better maintainability

---

# 6. Inheritance

Inheritance allows one class to inherit properties and methods from another class.

### Types of Inheritance

* Single Inheritance
* Multiple Inheritance
* Multilevel Inheritance
* Hierarchical Inheritance

### Example

```python
class Employee:

    def work(self):
        print("Employee working")

class Manager(Employee):

    def manage(self):
        print("Managing team")

obj = Manager()

obj.work()
obj.manage()
```

### Output

```text
Employee working
Managing team
```

### Advantages

* Code reusability
* Less duplication
* Better scalability

---

# 7. Polymorphism

Polymorphism means **one method with different behaviors**.

Different classes can have methods with the same name but different implementations.

### Example

```python
class Dog:

    def sound(self):
        print("Dog barks")

class Cat:

    def sound(self):
        print("Cat meows")

d = Dog()
c = Cat()

d.sound()
c.sound()
```

### Output

```text
Dog barks
Cat meows
```

### Benefits

* Flexible code
* Easy extension
* Improved readability

---

# 8. Abstraction

Abstraction means hiding implementation details and showing only essential features.

Python provides abstraction using the `abc` module.

### Example

```python
from abc import ABC, abstractmethod

class Shape(ABC):

    @abstractmethod
    def area(self):
        pass


class Rectangle(Shape):

    def __init__(self, length, breadth):
        self.length = length
        self.breadth = breadth

    def area(self):
        return self.length * self.breadth


obj = Rectangle(5, 4)

print(obj.area())
```

### Output

```text
20
```

### Benefits

* Better security
* Cleaner code
* Reduced complexity

---

# 9. Method Overriding

Method overriding occurs when a child class redefines a parent class method.

### Example

```python
class Animal:

    def sound(self):
        print("Animal makes sound")

class Dog(Animal):

    def sound(self):
        print("Dog barks")

obj = Dog()

obj.sound()
```

### Output

```text
Dog barks
```

### Benefits

* Allows customized behavior
* Supports polymorphism

---

# 10. `super()` Function

The `super()` function is used to access methods and constructors of the parent class.

### Example

```python
class Parent:

    def __init__(self):
        print("Parent Constructor")

class Child(Parent):

    def __init__(self):
        super().__init__()
        print("Child Constructor")

obj = Child()
```

### Output

```text
Parent Constructor
Child Constructor
```

### Purpose of `super()`

* Calls parent constructor
* Avoids duplicate code
* Supports inheritance

---

# 11. Access Modifiers

Access modifiers define the accessibility of variables and methods.

## 1. Public Members

Accessible from anywhere.

### Example

```python
class Student:

    def __init__(self):
        self.name = "Aditya"

obj = Student()

print(obj.name)
```

### Output

```text
Aditya
```

---

## 2. Protected Members

Protected members are indicated using a single underscore `_`.

Accessible within class and derived class.

### Example

```python
class Parent:

    def __init__(self):
        self._name = "Python"

class Child(Parent):

    def display(self):
        print(self._name)

obj = Child()

obj.display()
```

### Output

```text
Python
```

---

## 3. Private Members

Private members use double underscore `__`.

They cannot be directly accessed outside the class.

### Example

```python
class Student:

    def __init__(self):
        self.__marks = 90

    def display(self):
        print(self.__marks)

obj = Student()

obj.display()
```

### Output

```text
90
```

````python
# 1. Student Management System
class Student:

    def __init__(self, name, roll, marks):
        self.name = name
        self.roll = roll
        self.marks = marks

    def calculate_grade(self):
        if self.marks >= 90:
            return "A"
        elif self.marks >= 75:
            return "B"
        else:
            return "C"

    def display_details(self):
        print("\nStudent Details")
        print("Name:", self.name)
        print("Roll No:", self.roll)
        print("Marks:", self.marks)
        print("Grade:", self.calculate_grade())


student1 = Student("Aditya", 101, 88)
student1.display_details()
````

````python
# 2. Bank Account System (Encapsulation)

class BankAccount:

    def __init__(self, balance):
        self.__balance = balance

    def deposit(self, amount):
        self.__balance += amount
        print("Deposited:", amount)

    def withdraw(self, amount):
        if amount <= self.__balance:
            self.__balance -= amount
            print("Withdrawn:", amount)
        else:
            print("Insufficient Balance")

    def get_balance(self):
        return self.__balance


account = BankAccount(5000)

account.deposit(2000)
account.withdraw(1000)

print("Current Balance:", account.get_balance())
````

````python
# 5. Shape Area Calculator (Abstraction)

from abc import ABC, abstractmethod


class Shape(ABC):

    @abstractmethod
    def area(self):
        pass


class Rectangle(Shape):

    def __init__(self, l, b):
        self.l = l
        self.b = b

    def area(self):
        return self.l * self.b


rectangle = Rectangle(5, 4)

print("\nArea of Rectangle:", rectangle.area())
````

````python
# 6. Library Management System (Mini Project)

class Book:

    def __init__(self, title):
        self.title = title
        self.is_issued = False


class Member:

    def __init__(self, name):
        self.name = name


class Library:

    def __init__(self):
        self.books = []

    def add_book(self, book):
        self.books.append(book)
        print(f'Book "{book.title}" added.')

    def issue_book(self, title):
        for book in self.books:
            if book.title == title and not book.is_issued:
                book.is_issued = True
                print(f'Book "{title}" issued.')
                return
        print("Book not available.")

    def return_book(self, title):
        for book in self.books:
            if book.title == title and book.is_issued:
                book.is_issued = False
                print(f'Book "{title}" returned.')
                return
        print("Book not found.")

    def display_available_books(self):
        print("\nAvailable Books:")
        available = False

        for book in self.books:
            if not book.is_issued:
                print("-", book.title)
                available = True

        if not available:
            print("No books available.")


library = Library()

book1 = Book("Python Basics")
book2 = Book("OOP in Python")
book3 = Book("Data Structures")

library.add_book(book1)
library.add_book(book2)
library.add_book(book3)

library.display_available_books()

library.issue_book("Python Basics")

library.display_available_books()

library.return_book("Python Basics")

library.display_available_books()
````

# OOP Interview Questions & Answers

## 1. What is a class?

A **class** is a blueprint or template used to create objects. It defines attributes (variables) and methods (functions) that an object will have.

### Example

```python
class Student:
    pass
```

---

## 2. What is an object?

An **object** is an instance of a class. It is created using a class and can access the class's attributes and methods.

### Example

```python
class Student:
    name = "Aditya"

student1 = Student()

print(student1.name)
```

---

## 3. Difference between Class and Object?

| Class                                          | Object                                |
| ---------------------------------------------- | ------------------------------------- |
| Blueprint or template                          | Instance of class                     |
| Does not occupy memory until object is created | Occupies memory                       |
| Used to define properties and methods          | Used to access properties and methods |

### Example

```python
class Car:
    pass

car1 = Car()
```

Here:

* `Car` → Class
* `car1` → Object

---

## 4. What is inheritance?

**Inheritance** is an OOP concept where one class acquires properties and methods of another class.

The parent class is called the **base class**, and the child class is called the **derived class**.

### Example

```python
class Employee:

    def work(self):
        print("Employee working")

class Manager(Employee):

    def manage(self):
        print("Managing team")
```

---

## 5. What is polymorphism?

**Polymorphism** means **one method having multiple forms**. Different classes can use the same method name with different behavior.

### Example

```python
class Car:

    def start(self):
        print("Car starts")

class Bike:

    def start(self):
        print("Bike starts")
```

---

## 6. What is encapsulation?

**Encapsulation** means wrapping data and methods together in a single unit and restricting direct access to data.

It is used for **data security**.

### Example

```python
class Bank:

    def __init__(self):
        self.__balance = 5000
```

Here, `__balance` is private.

---

## 7. What is abstraction?

**Abstraction** means hiding unnecessary implementation details and showing only essential functionality.

Python uses the `abc` module for abstraction.

### Example

```python
from abc import ABC, abstractmethod

class Shape(ABC):

    @abstractmethod
    def area(self):
        pass
```

---

## 8. Difference between Method Overriding and Method Overloading?

| Method Overriding                     | Method Overloading                         |
| ------------------------------------- | ------------------------------------------ |
| Same method in parent and child class | Same method name with different parameters |
| Achieved using inheritance            | Achieved by changing arguments             |
| Runtime polymorphism                  | Compile-time concept                       |

### Method Overriding Example

```python
class Animal:

    def sound(self):
        print("Animal sound")

class Dog(Animal):

    def sound(self):
        print("Dog barks")
```

### Method Overloading Example

```python
class Math:

    def add(self, a, b, c=0):
        return a + b + c
```

## 9. What is the purpose of `super()`?

The `super()` function is used to call methods or constructors of the parent class.

It helps avoid code duplication.

```python
class Child(Parent):

    def __init__(self):
        super().__init__()
        print("Child Constructor")
```
